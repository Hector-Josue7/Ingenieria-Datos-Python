```python
import sqlalchemy as db 
import pandas as pd 
```

```python
#engine = db.create_engine('mysql+mysqlconnector://root:root@localhost:3306/retail_db') # Cadena de conexion a MySQL
engine = db.create_engine('mysql+mysqlconnector://root:root@localhost:3310/retail_db')
try:
    conexion = engine.connect() # aqui se establece la conexion a la base de datos
    print("Conexión exitosa")
except Exception as e:
    print("Error al conectar:", e)
```

```python
df_departments = pd.read_sql_table('departments', conexion) # Si se lee la tabla completa 
df_departments
```

```python
df_departments = pd.read_sql_query("SELECT * FROM departments", conexion) 

df_departments # imprimimos 
```

```python
df_departments = pd.read_sql("SELECT * FROM departments", conexion)

df_departments
```
