```python
import pandas as pd 
from azure.storage.filedatalake import DataLakeServiceClient
from io import StringIO # almacenar en bufer los datos que vengan de ese archivo 
```

```python
storage_account ="sakrm001"
sas_token =''
container = 'raw'
file_name = 'departments.csv' # estas 4 variables me van a servir para conectarme al data lake y extraer ese archivo 
```

```python
def obtener_cliente_adls(): # esta funcion va a devolver el cliente del data lake  
    client = DataLakeServiceClient(account_url=f"https://{storage_account}.dfs.core.windows.net", credential=sas_token) # uri del datalake  
    return client
```

```python
client = obtener_cliente_adls() # se debe retornar una conexion hacia el data lake 
container_client = client.get_file_system_client(container) # file_system=container
file = container_client.get_file_client(file_name) # obtener el archivo o la referencia al archivo 
```

```python
download = file.download_file()
download_bytes = download.readall() 
file_content = download_bytes.decode('utf-8') # convertir los bytes a string
file_content
```

```python
df = pd.read_csv(StringIO(file_content)) 

df  
```
